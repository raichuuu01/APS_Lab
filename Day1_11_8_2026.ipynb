```python
head=0
tail=0
import random
for i in range(100):
  num=random.randint(1,2)
  if num==1:
    head+=1
  else:
    tail+=1
print("number of heads: ",head)
print("number of tails: ", tail)
print("experimental probability of heads: ",head/100)
print("experimental probability of tails: ", tail/100)

#this almost matches with the expected probability of the chance of heads and tails having probability 0.5 each
```

```python
print("Die rolled 6000 times")
from collections import defaultdict
import random

dic=defaultdict(int)

for i in range(6000):
  num=random.randint(1,6)
  dic[num]+=1
print("occurence of number: ",dic)
print("occurence of 1",dic[1]/6)
print("occurence of 2",dic[2]/6)
print("occurence of 3",dic[3]/6)
print("occurence of 4",dic[4]/6)
print("occurence of 5",dic[5]/6)
print("occurence of 6",dic[6]/6)

#the results show that each number on the die occurs roughly the same number of timess
```

```python
print("Addition rule for disjoint events ")
from collections import defaultdict
import random

dic=defaultdict(int)

for i in range(10000):
  num=random.randint(1,6)
  dic[num]+=1
print("occurence of A",dic[1]/10000)
print("occurence of B",dic[6]/10000)
print("AUB: ",(dic[1]+dic[6])/10000)

#probability of event A and event B adds up to the probability of A ∪ B
#the two events are disjoint and cannot happen at the same time
#the results are close to the expected valueswith small differences due to randomness
```

```python
print("Addition rule for overlapping events")

from collections import defaultdict
import random

dic = defaultdict(int)

for i in range(10000):
    num = random.randint(1, 6)
    dic[num] += 1

a = (dic[2] + dic[4] + dic[6]) / 10000

b = (dic[4] + dic[5] + dic[6]) / 10000


aintersectionb = (dic[4] + dic[6]) / 10000


aunionb = (dic[2] + dic[4] + dic[5] + dic[6]) / 10000

print("P(A) =", a)
print("P(B) =", b)
print("P(A intersection B) =", aintersectionb)
print("P(A union B) =", aunionb)

result = a + b - aintersectionb

print("P(A) + P(B) - P(A intersection B) =", result)

#the experimental value of p(a∪b) is almost the same as p(a)+p(b)−p(a∩b)
# this shows that the addition rule works for overlapping events, where the common outcomes need to be subtracted once to avoid counting them twice
```
